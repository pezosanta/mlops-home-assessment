### Authenticate to Azure and connecting to Azure ML Workspace


```python
from pathlib import Path

from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential, InteractiveBrowserCredential

ws_configs_root_folder = Path.cwd().parent.joinpath("infra/azure/azureml/ws_configs")
ws_config = "continental-aml.json"

try:
    credential = DefaultAzureCredential()
    # Check if given credential can get token successfully.
    credential.get_token("https://management.azure.com/.default")
except Exception as ex:
    # Fall back to InteractiveBrowserCredential in case DefaultAzureCredential not work
    credential = InteractiveBrowserCredential()

ml_client = MLClient.from_config(credential=credential, path=ws_configs_root_folder.joinpath(ws_config))
```

### Submit Azure ML pipelines
Configure pipeline properties and inputs as they fit best. Hyperparameters can only be modified in the AML pipeline definition yaml (`infra/azure/azureml/pipelines/training/pipeline.yaml`).


```python
from azure.ai.ml import load_job

pipeline_job_config = Path.cwd().parent.joinpath("infra/azure/azureml/pipelines/training/pipeline.yaml")

pipeline_job = load_job(pipeline_job_config)

experiment_name = "Continental-Demo"

pipeline_job.inputs.batch_size = 256
pipeline_job.inputs.num_workers = 8
pipeline_job.inputs.prefetch_factor = 4
pipeline_job.inputs.num_epochs = 10
pipeline_job.inputs.dev_mode = False

ml_client.jobs.create_or_update(
    pipeline_job, 
    experiment_name=experiment_name,
)
```
